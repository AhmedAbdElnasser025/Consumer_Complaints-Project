```python
df = pd.read_csv('consumer_complaints.csv')
pd.set_option('display.max_columns',None)
df.head(7)
```

```python
df.info()
```

```python
df['date_received'] = pd.to_datetime(df['date_received'])
df['date_sent_to_company'] = pd.to_datetime(df['date_sent_to_company'])
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df.drop(columns=[
    'sub_product',
    'sub_issue',
    'consumer_complaint_narrative',
    'company_public_response',
    'zipcode',
    'tags',
    'consumer_consent_provided'
], inplace=True)
```

```python
df.isna().sum()
```

```python
df.dropna(subset=['state'] , inplace=True)
```

```python
df.isna().sum()
```

```python
df.duplicated().sum()
```

```python
check_list = ['product','issue','company'
              ,'state','submitted_via','company_response_to_consumer'
              ,'timely_response','consumer_disputed?']

def unique_value_check(df,col_name):
    '''Unique values check'''
    x = df[col_name].value_counts(dropna= False)
    print(x)

for col_name in check_list :
    print(col_name)
    unique_value_check(df,col_name)
    print('\n')
```

```python
df['response_status'] = df['company_response_to_consumer'].map({
    'Closed with monetary relief': 'Resolved',
    'Closed with non-monetary relief': 'Resolved',
    'Closed with relief': 'Resolved',
    'Closed with explanation': 'Unresolved',
    'Closed without relief': 'Unresolved',
    'Closed': 'Unresolved',
    'Untimely response': 'Unresolved',
    'In progress': 'In Progress'
})
```
